```python
import numpy as np
import matplotlib.pyplot as plt
import random
from math import exp
import time as t
import statistics as s

def log_map(v, r, q):
    if v <= r:
        return (((-q / (r * r)) * ((r - v) * (r - v))) + q)
    else:
        return (((-q / ((1 - r) * (1 - r))) * ((r - v) * (r - v))) + q)

def sim_mapa(params, iterations):
    mu, mu1, nu, nu1, alpha = params
    x0 = 0.03
    y0 = 0.01
    x = [x0]
    y = [y0]
    for i in range(1, iterations):
        xn = (log_map(x[i - 1], nu, mu) + (alpha * (y[i - 1] - x[i - 1]))) % 1
        yn = (log_map(y[i - 1], nu1, mu1) + (alpha * (x[i - 1] - y[i - 1]))) % 1
        x.append(xn)
        y.append(yn)
    #Notar que se simula a partir del segundo punto. De esta forma, un atacante puede trabajar sin condiciones iniciales
    return np.column_stack((x[1:], y[1:])) 
    

def objective_function(params):
    iterations = 30 #Menos puntos y no se puede identificar el sistema, más puntos y se congregan muchos puntos en el mismo lugar
    estimated_output = sim_mapa(params,iterations)
    error = 1 / iterations * np.sum((true_output - estimated_output) ** 2)
    return error

#Se trabaja la graficación como una función
def graficar_mapas(true_output, estimated_output):
    plt.figure(figsize=(7, 6))
    plt.scatter(true_output[:, 0], true_output[:, 1], s=1, color='blue', label='Mapa Original')
    plt.scatter(estimated_output[:, 0], estimated_output[:, 1], s=1, color='red', label='Mapa Aproximado')
    plt.title('Mapa Original vs Aproximado')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend(markerscale=5)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

    distances = np.linalg.norm(true_output - estimated_output, axis=1)
    avg_distance = np.mean(distances)
    print(f'Distancia promedio entre puntos correspondientes: {avg_distance:.6e}')
    return avg_distance

#Inicialización caótica para las partículas
def generate_logistic_positions(num_particles, x0=0.6, mu=4.0):
    x = x0
    positions = []
    for _ in range(num_particles):
        x = mu * x * (1 - x)
        positions.append(x)
    return positions

# Particle class
#Nathan Rooy en Github tiene una implementación básica de PSO, y utilicé su código como plantilla para generar las clases
class Particle:
    def __init__(self, bounds, logistic_positions, particle_id):
        global num_dimensions
        self.position_i = [
            bounds[i][0] + (bounds[i][1] - bounds[i][0]) * logistic_positions[particle_id] 
            for i in range(num_dimensions)
        ]
        self.velocity_i = [random.uniform(-1, 1) for _ in range(num_dimensions)] 
        self.pos_best_i = []  # Best individual position
        self.err_best_i = -1  # Best individual error
        self.err_i = -1  # Current error

    def evaluate(self, costFunc):
        self.err_i = costFunc(self.position_i)
        if self.err_i < self.err_best_i or self.err_best_i == -1:
            self.pos_best_i = list(self.position_i)
            self.err_best_i = self.err_i

    def update_velocity(self, pos_best_g, w):
        c1, c2 = 2,2 #Correspondiendo a lo utilizado para CBPSO2
        for i in range(num_dimensions):
            r1, r2 = random.random(), random.random()
            vel_cognitive = c1 * r1 * (self.pos_best_i[i] - self.position_i[i])
            vel_social = c2 * r2 * (pos_best_g[i] - self.position_i[i])
            self.velocity_i[i] = w * self.velocity_i[i] + vel_cognitive + vel_social

    def update_position(self, bounds):
        for i in range(num_dimensions):
            self.position_i[i] += self.velocity_i[i]
            self.position_i[i] = np.clip(self.position_i[i], bounds[i][0], bounds[i][1])

    def chaotic_search(self, global_best, a, bounds): 
        mu = 4.0
        x = random.uniform(0.1, 0.9)
        for _ in range(100):
            x = mu * x * (1 - x)
        for i in range(num_dimensions):
            search_range = a * global_best[i]
            new_pos = global_best[i] - search_range + (2 * search_range) * x
            self.position_i[i] = np.clip(new_pos, bounds[i][0], bounds[i][1])

class CBPSO:
    def __init__(self, costFunc, bounds, num_particles, maxiter):
        global num_dimensions
        num_dimensions = len(bounds)
        self.err_best_g = float("inf")  # Global best error
        pos_best_g = []  # Global best position
        a = 0.1  # Chaotic search range factor


        logistic_positions = generate_logistic_positions(num_particles) #Inicialización caótica de cada partícula
        cambio_err = []
        cambio_iter = []
        
        #Se inicializa el swarm
        swarm = [Particle(bounds, logistic_positions, i) for i in range(num_particles)]

        for i in range(maxiter):
            w_max, w_min, p = 0.9, 0.2, 100
            w = (w_max - w_min) * exp(-p * (i / maxiter) ** 2) + w_min #inertia weight como función exponencial de las iteraciones

            for particle in swarm:
                particle.evaluate(costFunc)
                if particle.err_i < self.err_best_g:
                    pos_best_g = list(particle.position_i)
                    self.err_best_g = particle.err_i

            for particle in swarm:
                particle.update_velocity(pos_best_g, w)
                particle.update_position(bounds)

            for particle in swarm:
                if self.err_best_g > 1:
                    particle.chaotic_search(pos_best_g, a, bounds)
                elif 10 ** -(i + 2) < self.err_best_g < 10 ** -i:
                    particle.chaotic_search(pos_best_g, a / 2, bounds)

            cambio_iter.append(i+1)
            cambio_err.append(self.err_best_g)
            

        self.best_position = [float(param) for param in pos_best_g]
        print('FINAL:')
        print('Parámetros optimizados:', self.best_position)
        print('Función objetivo ECM optimizada:', self.err_best_g)
       #Graficacion de ECM vs iteraciones. No se usa la función graficar_mapas porque son para otro tipo de graficas
        plt.figure(figsize=(12, 5))
        plt.subplot(1, 2, 1)
        plt.plot(cambio_iter, cambio_err, color='blue', marker='o', markersize=2, linewidth=1)
        plt.title('ECM vs Iteraciones')
        plt.xlabel('Iteración')
        plt.ylabel('ECM')
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.show()
    

#--- RUN ---#
print("Se realiza optimización de CBPSO...")
true_params = [0.91, 0.85, 0.9, 0.333, 0.3314]
true_output = sim_mapa(true_params,30)
true_output_graph = sim_mapa(true_params,3000)
bounds = [(0.5, 1), (0.5, 1), (0.5, 1), (0.01, 1), (0.01, 0.5)]
bounds_iniciales = [(0.01,0.1),(0.01,0.1)]

ciclos = 100
besti = 0
bestmse = 10
cond_exito = 9e-3
exitos = 0
temp_avg = []
avg_err = []
dist_ciclos = []  # <-- nueva

for i in range(ciclos):
    print(f"Ciclo {i + 1}")
    print(f'Valores de x1 y x2 originales: {true_params}')
    temp_inicial = t.time()
    pso = CBPSO(objective_function, bounds, num_particles=1000, maxiter=250)
    temp_final = t.time()
    temp_neto = float(temp_final-temp_inicial)
    avg_err.append(pso.err_best_g)
    temp_avg.append(temp_neto)
    estimated_output = sim_mapa(pso.best_position,3000)
    
    d = graficar_mapas(true_output_graph, estimated_output)  # <-- capturar
    dist_ciclos.append(d)                                     # <-- guardar
    
    if pso.err_best_g <= bestmse:
        bestmse = pso.err_best_g
        besti = i
    if bestmse <= cond_exito:
        exitos += 1

avg_final = s.mean(avg_err)#sum(avg_err) / len(avg_err)
stdev_final = s.stdev(avg_err)

print(f"\n{'='*50}")
print(f"Resultados después de {ciclos} ciclos")
print(f"{'='*50}")

print(f"\n--- Distancia promedio (Primer proceso: obtención de parámetros) ---")
print(f"  Promedio : {s.mean(dist_paso1):.6e}")
print(f"  Desv. estándar: {s.stdev(dist_paso1):.6e}")
print(f"  Mínimo   : {min(dist_paso1):.6e}")
print(f"  Máximo   : {max(dist_paso1):.6e}")

print(f"\n--- Distancia promedio (Segundo proceso: conds. iniciales) ---")
print(f"  Promedio : {s.mean(dist_paso2):.6e}")
print(f"  Desv. estándar: {s.stdev(dist_paso2):.6e}")
print(f"  Mínimo   : {min(dist_paso2):.6e}")
print(f"  Máximo   : {max(dist_paso2):.6e}")

print(f"\nEl mejor ECM se dio en el ciclo {besti + 1}, igual a: {bestmse:.6e}")
print(f"Éxitos (ECM ≤ {cond_exito}): {exitos}")
print(f"Tiempo promedio por ciclo: {np.mean(temp_avg):.2f}s (σ = {np.std(temp_avg):.2f}s)")
```
